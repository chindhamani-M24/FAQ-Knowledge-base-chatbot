```python
# ==========================================
# Interactive FAQ & Knowledge Base Chatbot
# Google Colab Copy-Paste Ready Script
# ==========================================

import subprocess
import sys
import pandas as pd

def install_package(package):
    try:
        __import__(package)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", package])

print("--- Checking and installing required libraries ---")
install_package("pandas")
install_package("scikit-learn")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Knowledge Base Dataset (Questions & Answers)
faqs = [
    {
        "question": "What are your shipping times?",
        "answer": "Standard shipping takes 3-5 business days. Express shipping takes 1-2 business days."
    },
    {
        "question": "How can I return an item?",
        "answer": "You can initiate a return within 30 days of purchase through your account dashboard or by contacting support."
    },
    {
        "question": "What payment methods do you accept?",
        "answer": "We accept major credit cards (Visa, MasterCard, Amex), PayPal, and Apple Pay."
    },
    {
        "question": "How do I reset my password?",
        "answer": "Click on 'Forgot Password' on the login page and follow the instructions sent to your registered email."
    },
    {
        "question": "Are there any subscription fees?",
        "answer": "Our basic tier is free forever! Premium features are available starting at $9.99/month."
    },
    {
        "question": "How do I contact customer support?",
        "answer": "You can reach us via live chat on our website or by emailing support@example.com."
    }
]

df = pd.DataFrame(faqs)

print("\n--- Initializing NLP Knowledge Base & Vectorizer ---")
# Vectorize the FAQ questions using TF-IDF
vectorizer = TfidfVectorizer(stop_words='english')
question_vectors = vectorizer.fit_transform(df['question'])

# 2. Interactive Loop for Google Colab
print("\n" + "="*55)
print("🤖 INTERACTIVE FAQ CHATBOT READY")
print("Ask a question about our services below.")
print("Type 'exit' or 'quit' to stop.")
print("="*55 + "\n")

while True:
    try:
        user_query = input("You: ").strip()
    except EOFError:
        break

    if user_query.lower() in ['exit', 'quit']:
        print("\nChatbot: Goodbye! Have a great day!")
        break

    if not user_query:
        print("Chatbot: Please type a valid question.\n")
        continue

    # Vectorize user query and compute cosine similarity against knowledge base
    query_vector = vectorizer.transform([user_query])
    similarity_scores = cosine_similarity(query_vector, question_vectors)

    # Find the best match
    best_match_idx = similarity_scores.argmax()
    max_score = similarity_scores[0, best_match_idx]

    # Set a confidence threshold to filter out irrelevant queries
    if max_score < 0.15:
        print("Chatbot: I'm sorry, I couldn't find an exact answer to that. Please try rephrasing or contact support@example.com.\n")
    else:
        matched_question = df.iloc[best_match_idx]['question']
        matched_answer = df.iloc[best_match_idx]['answer']
        print(f"Chatbot (Matched FAQ: '{matched_question}' | Match Score: {max_score:.2f})")
        print(f"👉 {matched_answer}\n")
```
